```python
from ucimlrepo import fetch_ucirepo
import pandas as pd
import numpy as np
```

```python
adult = fetch_ucirepo(id=2)
X = adult.data.features
y = adult.data.targets
df = X.join(y)
```

```python
def clean_df(df):
    
    df = df.replace('?', 'Unknown')
    df = df.drop(columns=['native-country', 'fnlwgt', 'education'])
    
    df['income'] = df['income'].str.rstrip('.')

    for col in df.columns:
        if df[col].dtype.name in ['string', 'O', 'object']:
            df[col] = df[col].str.strip().fillna('Unknown')

    return df
```

```python
final_df = clean_df(df)
final_df.head(20)
```
