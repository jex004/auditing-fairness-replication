```python
#FINAL CLEANING FUNCTION FOR THE DF 

def clean_df(df):
    
    df = df.replace('?', 'Unknown')
    df = df.drop(columns=['native-country', 'fnlwgt', 'education'])
    
    df['income'] = df['income'].str.rstrip('.')

    for col in df.columns:
        if df[col].dtype.name in ['string', 'O', 'object']:
            df[col] = df[col].str.strip().fillna('Unknown')

    return df
```

```python
test = clean_df(df)
test.head(20)
```

```python
test.shape[0] == df.shape[0]
```

```python
#CHECKING TO SEE IF CAPITAL GAINS COLUMN IS REDUNDANT
```

```python
test.groupby('income')[['capital-gain']].mean()
```

```python
#very low variance, 91.7% of data points have capital gain of zero

test[test['capital-gain'] == 0]['capital-gain'].shape[0] / test.shape[0]
```

```python
#testing correlation using Point-Biserial Correlation ---> No correlation at all with high confidence 
import scipy
from scipy.stats import pointbiserialr

test['income_binary'] = test['income'].apply(lambda x: 1 if '>50k' in str(x).lower() else 0)

correlation, p_value = pointbiserialr(test['income_binary'], test['capital-gain'])

print(f"Correlation: {correlation:.4f}")
print(f"P-value: {p_value}")
```
